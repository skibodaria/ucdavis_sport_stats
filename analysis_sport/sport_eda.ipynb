## Explore the NOC dataset:

```python
# read the second file on NOC:
df_noc = pd.read_csv('../data_sport/noc_regions.csv')

# check the first two rows:
df_noc.head(2)
```

```python
# check the columns:
df_noc.columns
```

```python
df_noc.columns = df_noc.columns.str.strip().str.lower()
```

```python
# check the data types and null values:
df_noc.info()
```

```python
# check for duplicates:
sum(df_noc.duplicated())
# --> there are zero duplicates
```

```python
# in bash: pip install pandaserd
from pandaserd import ERD
# more info: https://pypi.org/project/pandaserd/ 

# initialize an ERD:
my_erd = ERD()

# add the dataframes:
my_erd.add_table(df, table_name = 'athletes', bg_color = 'pink')
my_erd.add_table(df_noc, table_name = 'regions', bg_color = 'lightblue')

# make relationships:
my_erd.create_rel('athletes', 'regions', on='noc', right_cardinality = '.')

my_erd.write_to_file('erd_output.txt')

# than the text file can be displayed using: https://edotor.net/
```

```python
df.columns
```

```python
df_athlete = df[['id', 'name', 'sex', 'age', 'height', 'weight', 'team', 'games']]
df_event = df[['event','sport','city', 'season', 'year', 'games']]
df_games = df[['year', 'games', 'noc']]
df_awards = df[['id', 'year','event', 'medal', 'sport','games']]
```

```python
df_event.head()
```

```python
df_athlete.head()
```

```python
df_games.head()
```

```python
df_awards.head()
```

```python
# initialize an ERD:
upd_erd = ERD()

# add the dataframes:
upd_erd.add_table(df_athlete, table_name = 'athletes', bg_color = 'pink')
upd_erd.add_table(df_noc, table_name = 'regions', bg_color = 'lightblue')
upd_erd.add_table(df_games, 'games', bg_color = 'lightyellow')
upd_erd.add_table(df_event, 'events', bg_color = 'grey')
upd_erd.add_table(df_awards, 'awards', bg_color = 'gold')


# make relationships:
upd_erd.create_rel('athletes', 'awards', on='id', right_cardinality = '*')
upd_erd.create_rel('games', 'events', on='games', right_cardiality = '*')
upd_erd.create_rel('games', 'regions', on='noc', right_cardiality = '.')
upd_erd.create_rel('events', 'awards', on='event', right_cardiality = '.')
upd_erd.create_rel('athletes', 'games', on='games', right_cardiality = '.')


upd_erd.write_to_file('erd_updated.txt')

# than the text file can be displayed using: https://edotor.net/
```
